```python
print("🌍 AI Travel Agent")

name = input("What is your name? ")
destination = input("Where do you want to travel? ")
budget = int(input("What is your budget ($)? "))

print("\n✈️ Travel Plan")
print(f"Hello {name}!")
print(f"Your destination is {destination}.")
print(f"Your budget is ${budget}.")

if budget < 500:
    print("💡 Suggestion: Choose budget hotels and economy flights.")
elif budget < 1500:
    print("💡 Suggestion: Choose a 3-star hotel and economy flight.")
else:
    print("💡 Suggestion: You can consider luxury hotels and premium flights.")

print("Have a safe and wonderful journey! 🌎")
```
